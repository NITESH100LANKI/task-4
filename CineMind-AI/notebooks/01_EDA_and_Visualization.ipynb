# CineMind AI: Advanced EDA & Visual Analytics
This notebook demonstrates exploratory data analysis on the MovieLens dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set paths
DATA_DIR = Path('../data/processed')
movies_df = pd.read_parquet(DATA_DIR / 'movies.parquet')
ratings_df = pd.read_parquet(DATA_DIR / 'ratings.parquet')

print('Movies:', movies_df.shape)
print('Ratings:', ratings_df.shape)
```

```python
# Genre Distribution
plt.figure(figsize=(12, 6))
genres = movies_df['genres'].str.split(' ', expand=True).stack().value_counts()
sns.barplot(x=genres.values, y=genres.index, palette='viridis')
plt.title('Global Genre Distribution')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```
